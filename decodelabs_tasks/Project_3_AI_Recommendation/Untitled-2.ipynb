```python
# Project 3: AI Recommendation Logic
# DecodeLabs Artificial Intelligence Training

print("=" * 60)
print("          AI RECOMMENDATION SYSTEM")
print("=" * 60)

print("\nWelcome to the AI Recommendation System!")
print("Tell us what you are interested in.")
print("You can choose from: action, comedy, science, technology,")
print("romance, adventure, education, music, sports")
print()

# Recommendation database
items = [
    {
        "name": "The Matrix",
        "category": "movie",
        "interests": ["action", "science", "technology"]
    },
    {
        "name": "Interstellar",
        "category": "movie",
        "interests": ["science", "adventure", "technology"]
    },
    {
        "name": "The Hangover",
        "category": "movie",
        "interests": ["comedy"]
    },
    {
        "name": "The Notebook",
        "category": "movie",
        "interests": ["romance"]
    },
    {
        "name": "Avengers: Endgame",
        "category": "movie",
        "interests": ["action", "adventure"]
    },
    {
        "name": "Python Programming",
        "category": "course",
        "interests": ["technology", "education"]
    },
    {
        "name": "Data Science Fundamentals",
        "category": "course",
        "interests": ["science", "technology", "education"]
    },
    {
        "name": "Football Highlights",
        "category": "sports",
        "interests": ["sports", "adventure"]
    },
    {
        "name": "Music Playlist",
        "category": "music",
        "interests": ["music", "romance"]
    }
]

# Get user preferences
user_input = input("Enter your interests separated by commas: ")

# Clean and process the input
user_preferences = []

for interest in user_input.lower().split(","):
    interest = interest.strip()

    if interest:
        user_preferences.append(interest)

print("\nYour interests:", ", ".join(user_preferences))

# Calculate similarity scores
recommendations = []

for item in items:
    score = 0

    for preference in user_preferences:
        if preference in item["interests"]:
            score += 1

    if score > 0:
        recommendations.append((item, score))

# Sort recommendations by similarity score
recommendations.sort(key=lambda x: x[1], reverse=True)

# Display recommendations
print("\n" + "=" * 60)
print("              RECOMMENDATIONS")
print("=" * 60)

if recommendations:
    for item, score in recommendations:
        print(f"\n{item['name']}")
        print(f"Type: {item['category']}")
        print(f"Matched Interests: {score}")
        print(f"Tags: {', '.join(item['interests'])}")
else:
    print("\nSorry, no matching recommendations were found.")
    print("Try interests such as:")
    print("action, comedy, science, technology, romance,")
    print("adventure, education, music, or sports.")

print("\nThank you for using the AI Recommendation System!")
```
